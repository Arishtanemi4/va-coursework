# 2021 Data Read and clean trial

```python
import pandas as pd

df = pd.read_excel("../data/econimic_activity_raw_2021_ts066.xlsx", sheet_name="Data")
```

```python
df
```

```python
header_mask = df.iloc[:, 0].astype(str).str.lower().str.startswith('local')

if header_mask.any():
    header_idx = header_mask.idxmax()
    df = df.iloc[header_idx:]
else:
    print(f"'local' not found in column 0 Falling back to row 7.")
    df = df.iloc[7:]
```

```python
df
```

```python
df.columns = df.iloc[0]
df = df.reset_index(drop=True)
df = df[1:]
```

```python
df
```

```python
df.columns = df.columns.str.lower()  # lowering col names (personal preferrence)
# regex to clean col names
df.columns = (df.columns.str.replace(r'[ /]', '_', regex=True).str.replace(r'[:()\-]', '', regex=True))
df = df.rename(columns={df.columns[1]:"mnemonics"})
df = df.rename(columns={df.columns[0]:"area"})

# Remove the las disclaimer and nan row
first_nan_idx = df[df['area'].isna()].index[0]
df = df.loc[:first_nan_idx - 1]
```

```python
df
```

# Changed Districts analysis

```python
df_2011 = pd.read_csv("../data/clean/joins/2011.csv")
df_2021 = pd.read_csv("../data/clean/joins/2021.csv")
```

```python
# Select only the columns that we need to understand the change
df_2011_area = df_2011[["area", "mnemonics"]]
df_2021_area = df_2021[["area", "mnemonics"]]
```

```python
join_keys = ['area']  # Join on area name rather than code. I mean ofc, names are eaier to track.

# First find the ones that are in 2011 but not in 2021. We know there were more in 2011 than in 2021 simply by the number of rows in each file.
dist_df = pd.merge(df_2011_area, df_2021_area, on=join_keys, how='outer')
dist_df.head()
```

```python
# look of nulls in mnemonics_y col as those are the ones not in 2021
missing_2021_df = dist_df[dist_df["mnemonics_y"].isna()]
missing_2021_df.head()
```

```python
# Similarly find ones in 2021 but not in 2011.
dist_2_df = pd.merge(df_2011_area, df_2021_area, on=join_keys, how='right')
dist_2_df.head()
```

```python
# Same as 2011 but with otjher column instead
missing_2011_df = dist_2_df[dist_2_df["mnemonics_x"].isnull()]
missing_2011_df.head()
```

```python
# tad da da we have our list of changed districts lets get deeper now
missing_2011_df = missing_2011_df.sort_values(by="mnemonics_y")
missing_2011_df
```

```python
missing_2021_df = missing_2021_df.sort_values(by="mnemonics_x")
missing_2021_df
```

Now all we need to do is fidn out whcih county from the `missing_2021_df` was merged into which counties.

[[Conversion LAD 2011 to LAD 2021](https://ckan.publishing.service.gov.uk/dataset/lsoa-2011-to-lsoa-2021-to-local-authority-district-2022-exact-fit-lookup-for-ew-v3/resource/59d599eb-43e1-4ba3-90ac-301a7e4e4da0)]: https://ckan.publishing.service.gov.uk/dataset/lsoa-2011-to-lsoa-2021-to-local-authority-district-2022-exact-fit-lookup-for-ew-v3/resource/59d599eb-43e1-4ba3-90ac-301a7e4e4da0

```python
df_map = pd.read_csv('../data/clean/joins/LAD_Mapping_2011_to_2021.csv')

df_2011_mapped = pd.merge(df_2011, df_map, left_on='mnemonics', right_on='LAD2011', how='left')
df_2011_mapped['mnemonics'] = df_2011_mapped['LAD2021'].fillna(df_2011_mapped['mnemonics'])
df_2011_mapped['area'] = df_2011_mapped['LAD2021NM'].fillna(df_2011_mapped['area'])
df_2011_mapped = df_2011_mapped.drop(columns=['LAD2011NM', 'LAD2011', 'LAD2021NM', 'LAD2021'])
df_2011_agg = df_2011_mapped.groupby(['area', 'mnemonics']).sum(numeric_only=True).reset_index()
```
